```python
#  Python Exception Handling

# ============================================================
# PYTHON EXCEPTION HANDLING - TOP 20 SIMPLE PROGRAMS
# ============================================================
#
# Exception Handling ka use program mein errors ko handle
# karne ke liye hota hai.
#
# Main keywords:
# try     -> risky code likhte hain
# except  -> error aane par ye code chalta hai
# else    -> agar error nahi aaya to ye chalta hai
# finally -> error aaye ya na aaye, ye normally chalta hai
# raise   -> manually error generate karne ke liye
#
# ============================================================


# ============================================================
# PROGRAM 1: BASIC TRY-EXCEPT
# ============================================================

try:
    # Yahan error generate hoga kyunki 10 ko 0 se divide
    # nahi kar sakte.
    result = 10 / 0

    print(result)

except:
    # Agar koi bhi error aayega to ye message print hoga.
    print("Kuch error aa gaya!")


# ============================================================
# PROGRAM 2: ZERO DIVISION ERROR
# ============================================================

try:
    # Do numbers create kar rahe hain.
    a = 10
    b = 0

    # Zero se division karne par ZeroDivisionError aayega.
    result = a / b

    print(result)

except ZeroDivisionError:
    # Sirf ZeroDivisionError ko handle kar rahe hain.
    print("Zero se divide nahi kar sakte!")


# ============================================================
# PROGRAM 3: VALUE ERROR
# ============================================================

try:
    # "hello" ko integer mein convert karne ki koshish.
    # Ye possible nahi hai, isliye ValueError aayega.
    age = int("hello")

    print(age)

except ValueError:
    # Invalid value hone par ye message print hoga.
    print("Valid number enter karo!")


# ============================================================
# PROGRAM 4: TYPE ERROR
# ============================================================

try:
    # Integer aur string ko directly add nahi kar sakte.
    # Isliye TypeError aayega.
    result = 10 + "20"

    print(result)

except TypeError:
    # Different data types ki problem ko handle kar rahe hain.
    print("Different data types ko directly add nahi kar sakte!")


# ============================================================
# PROGRAM 5: INDEX ERROR
# ============================================================

# Ek list create kar rahe hain.
numbers = [10, 20, 30]

try:
    # List mein sirf index 0, 1 aur 2 available hain.
    # Index 5 available nahi hai.
    print(numbers[5])

except IndexError:
    # Invalid index hone par ye message print hoga.
    print("List mein ye index exist nahi karta!")


# ============================================================
# PROGRAM 6: KEY ERROR
# ============================================================

# Dictionary create kar rahe hain.
student = {
    "name": "Rahul",
    "age": 20
}

try:
    # Dictionary mein "city" key available nahi hai.
    print(student["city"])

except KeyError:
    # Missing key hone par KeyError handle hoga.
    print("Ye key dictionary mein nahi hai!")


# ============================================================
# PROGRAM 7: FILE NOT FOUND ERROR
# ============================================================

try:
    # abc.txt naam ki file read mode mein open kar rahe hain.
    # Agar file exist nahi karti to FileNotFoundError aayega.
    file = open("abc.txt", "r")

    # File ka content read kar rahe hain.
    print(file.read())

    # File close kar rahe hain.
    file.close()

except FileNotFoundError:
    # File nahi milne par ye message print hoga.
    print("File nahi mili!")


# ============================================================
# PROGRAM 8: MULTIPLE EXCEPTIONS
# ============================================================

try:
    # User se first number le rahe hain.
    a = int(input("First number: "))

    # User se second number le rahe hain.
    b = int(input("Second number: "))

    # Division perform kar rahe hain.
    result = a / b

    print("Result:", result)

except ValueError:
    # Agar user number ki jagah text enter kare.
    print("Number hi enter karo!")

except ZeroDivisionError:
    # Agar second number 0 ho.
    print("Zero se divide nahi kar sakte!")


# ============================================================
# PROGRAM 9: MULTIPLE EXCEPTIONS IN ONE EXCEPT
# ============================================================

try:
    # User se number input le rahe hain.
    number = int(input("Number enter karo: "))

    # Number ko 10 se divide kar rahe hain.
    result = 10 / number

    print("Result:", result)

except (ValueError, ZeroDivisionError):
    # ValueError ya ZeroDivisionError dono ko
    # same except block mein handle kar rahe hain.
    print("Input ya calculation mein problem hai!")


# ============================================================
# PROGRAM 10: TRY-EXCEPT-ELSE
# ============================================================

try:
    # User se number input le rahe hain.
    number = int(input("Number enter karo: "))

except ValueError:
    # Agar input integer nahi hua to ye chalega.
    print("Invalid number!")

else:
    # Agar koi error nahi aaya to else chalega.
    print("Aapne enter kiya:", number)


# ============================================================
# PROGRAM 11: TRY-EXCEPT-FINALLY
# ============================================================

try:
    # Normal division kar rahe hain.
    number = 10 / 2

    print("Result:", number)

except ZeroDivisionError:
    # Agar zero se division hota to ye block chalta.
    print("Zero se divide nahi kar sakte!")

finally:
    # Error aaye ya na aaye, finally execute hota hai.
    print("Program complete!")


# ============================================================
# PROGRAM 12: TRY + EXCEPT + ELSE + FINALLY
# ============================================================

try:
    # User se integer input le rahe hain.
    number = int(input("Number enter karo: "))

except ValueError:
    # Invalid input hone par except chalega.
    print("Invalid number!")

else:
    # Error na hone par else chalega.
    print("Number hai:", number)

finally:
    # Finally har situation mein execute hota hai.
    print("Program finished!")


# ============================================================
# PROGRAM 13: GENERIC EXCEPTION
# ============================================================

try:
    # Zero se division kar rahe hain.
    # Error generate hoga.
    result = 10 / 0

    print(result)

except Exception as error:
    # Exception object ko variable "error" mein store kar rahe hain.
    print("Error:", error)


# ============================================================
# PROGRAM 14: INPUT VALIDATION
# ============================================================

try:
    # User se age le rahe hain.
    age = int(input("Age enter karo: "))

    # Check kar rahe hain ki age negative to nahi hai.
    if age < 0:
        print("Age negative nahi ho sakti!")

    else:
        # Valid age hone par age print hogi.
        print("Your age:", age)

except ValueError:
    # Agar user number ki jagah text enter kare.
    print("Age number mein enter karo!")


# ============================================================
# PROGRAM 15: CALCULATOR WITH EXCEPTION HANDLING
# ============================================================

def calculator():

    try:
        # User se first number le rahe hain.
        a = float(input("First number: "))

        # User se second number le rahe hain.
        b = float(input("Second number: "))

        # User se operator le rahe hain.
        operator = input("Operator (+ - * /): ")

        # Addition
        if operator == "+":
            print("Answer:", a + b)

        # Subtraction
        elif operator == "-":
            print("Answer:", a - b)

        # Multiplication
        elif operator == "*":
            print("Answer:", a * b)

        # Division
        elif operator == "/":
            print("Answer:", a / b)

        # Agar koi valid operator nahi diya.
        else:
            print("Invalid operator!")

    except ValueError:
        # Agar number ki jagah invalid value enter hui.
        print("Please valid numbers enter karo!")

    except ZeroDivisionError:
        # Agar second number zero hua.
        print("Zero se divide nahi kar sakte!")


# Calculator function ko call kar rahe hain.
calculator()


# ============================================================
# PROGRAM 16: FILE HANDLING WITH TRY-EXCEPT-FINALLY
# ============================================================

# Initially file variable ko None rakh rahe hain.
file = None

try:
    # File open kar rahe hain.
    file = open("data.txt", "r")

    # File ka content read kar rahe hain.
    content = file.read()

    # Content print kar rahe hain.
    print(content)

except FileNotFoundError:
    # Agar file nahi mili to ye message aayega.
    print("File nahi mili!")

finally:
    # Check kar rahe hain ki file actually open hui thi ya nahi.
    if file is not None:

        # File ko close kar rahe hain.
        file.close()

    # Ye message finally ki wajah se execute hoga.
    print("File operation complete!")


# ============================================================
# PROGRAM 17: RAISE KEYWORD
# ============================================================

# Example age.
age = 15

try:
    # Check kar rahe hain ki age 18 se kam hai ya nahi.
    if age < 18:

        # Hum manually ValueError create kar rahe hain.
        raise ValueError("Age 18 se kam hai!")

    # Agar age 18 ya usse zyada hai.
    print("Allowed")

except ValueError as error:
    # Manually generated error ko handle kar rahe hain.
    print("Error:", error)


# ============================================================
# PROGRAM 18: CUSTOM EXCEPTION
# ============================================================

# Apni custom exception class bana rahe hain.
# Ye Exception class se inherit kar rahi hai.
class AgeError(Exception):
    pass


try:
    # Example age.
    age = 15

    # Check kar rahe hain ki age 18 se kam hai.
    if age < 18:

        # Apni custom AgeError generate kar rahe hain.
        raise AgeError("Aapki age 18 se kam hai!")

except AgeError as error:
    # Custom exception ko handle kar rahe hain.
    print("Error:", error)


# ============================================================
# PROGRAM 19: NESTED TRY-EXCEPT
# ============================================================

try:

    # Outer try start.
    print("Outer try start")

    try:
        # Inner try ke andar division kar rahe hain.
        # ZeroDivisionError aayega.
        number = 10 / 0

        print(number)

    except ZeroDivisionError:
        # Inner error yahin handle ho jayega.
        print("Inner error handle ho gaya!")

    # Inner exception handle hone ke baad
    # outer try continue karega.
    print("Outer try continue")

except:
    # Agar outer try mein koi aur error aaye.
    print("Outer error")


# ============================================================
# PROGRAM 20: REAL-LIFE STUDENT RESULT PROGRAM
# ============================================================

def calculate_result():

    try:
        # User se marks input le rahe hain.
        marks = int(input("Marks enter karo: "))

        # Check kar rahe hain marks 0 se kam ya 100 se zyada
        # to nahi hain.
        if marks < 0 or marks > 100:

            # Invalid marks hone par manually ValueError raise.
            raise ValueError(
                "Marks 0 se 100 ke beech hone chahiye!"
            )

        # Agar marks 40 ya usse zyada hain.
        if marks >= 40:
            print("Pass")

        # Agar marks 40 se kam hain.
        else:
            print("Fail")

    except ValueError as error:
        # Invalid marks/input ko handle kar rahe hain.
        print("Error:", error)

    finally:
        # Result checking complete hone par ye message.
        print("Result checking complete!")


# Function ko call kar rahe hain.
calculate_result()


# ============================================================
# IMPORTANT EXCEPTION TYPES
# ============================================================
#
# ValueError
# -> Value galat ho, jaise int("hello")
#
# TypeError
# -> Wrong data types ka operation
#
# ZeroDivisionError
# -> Kisi number ko 0 se divide karna
#
# IndexError
# -> List/string ka invalid index
#
# KeyError
# -> Dictionary mein missing key
#
# FileNotFoundError
# -> File exist nahi karti
#
# NameError
# -> Undefined variable use karna
#
# AttributeError
# -> Object mein required attribute/method nahi hai
#
# ImportError
# -> Import mein problem
#
# ModuleNotFoundError
# -> Module nahi milta
#
# PermissionError
# -> Permission nahi hai
#
# OverflowError
# -> Number calculation limit se bahar chali jaye
#
# RuntimeError
# -> General runtime problem
#
# ============================================================


# ============================================================
# MASTER EXCEPTION HANDLING SYNTAX
# ============================================================

try:
    # Yahan risky code likho.
    pass

except ValueError:
    # ValueError handle karo.
    pass

except ZeroDivisionError:
    # ZeroDivisionError handle karo.
    pass

else:
    # Agar koi error nahi aaya to ye chalega.
    pass

finally:
    # Error aaye ya na aaye, ye chalega.
    pass


# ============================================================
# MANUALLY ERROR CREATE KARNA - RAISE
# ============================================================

try:

    # Example condition.
    age = 15

    # Agar age 18 se kam hai.
    if age < 18:

        # Manually error generate kar rahe hain.
        raise ValueError("Age 18 se kam hai!")

except ValueError as error:

    # Generated error ko handle kar rahe hain.
    print(error)


# ============================================================
# BEST PRACTICE
# ============================================================
#
# 1. Specific exception use karo:
#
#    except ValueError:
#
#    Ye better hai.
#
# 2. Sirf zarurat par generic Exception use karo:
#
#    except Exception as error:
#
# 3. Bare except avoid karo:
#
#    except:
#
#    Kyunki ye almost har error ko catch kar leta hai.
#
# 4. Error ko ignore mat karo:
#
#    except:
#        pass
#
#    Isse problem hide ho sakti hai.
#
# 5. finally ka use cleanup ke liye useful hai,
#    jaise file close karna.
#
# ============================================================
# END
# ============================================================
```
